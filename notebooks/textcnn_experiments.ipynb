## 1. Setup and Imports

````python
import os
import sys
import json
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import torch

# Add project root to path
sys.path.append(os.path.abspath('..'))

from utils.preprocessing import run_preprocessing_pipeline
from models.model_textcnn import TextCNN, count_parameters

print(f'PyTorch version : {torch.__version__}')
print(f'CUDA available  : {torch.cuda.is_available()}')
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f'Using device    : {DEVICE}')
````

## 2. Load and Preprocess Dataset

````python
DATA_PATH = '../data/reviews.csv'  # Update path as needed

data = run_preprocessing_pipeline(
    filepath    = DATA_PATH,
    sample_size = 50000,
    max_vocab   = 20000,
    max_seq_len = 200,
)

print(f"Vocabulary size : {data['vocab_size']:,}")
print(f"Num classes     : {data['num_classes']}")
print(f"Class labels    : {list(data['label_encoder'].classes_)}")
print(f"X_train shape   : {data['X_train'].shape}")
print(f"X_val shape     : {data['X_val'].shape}")
print(f"X_test shape    : {data['X_test'].shape}")
````

## 3. Model Architecture

````python
model = TextCNN(
    vocab_size   = data['vocab_size'],
    embed_dim    = 128,
    num_filters  = 128,
    kernel_sizes = [2, 3, 4],
    num_classes  = 3,
    dropout_rate = 0.5,
).to(DEVICE)

print(model)
print(f'\nTotal trainable parameters: {count_parameters(model):,}')
````

## 4. Train the Model

Run the training script from the terminal using:

```bash
python models/train_textcnn.py --data_path data/reviews.csv
```

The training script saves results to the `results/` folder automatically.

## 5. Load and Display Training Curves

````python
curves_path = '../results/textcnn_training_curves.png'

if os.path.exists(curves_path):
    img = plt.imread(curves_path)
    plt.figure(figsize=(12, 4))
    plt.imshow(img)
    plt.axis('off')
    plt.title('TextCNN — Training & Validation Curves')
    plt.tight_layout()
    plt.show()
else:
    print('Training curves not found. Run the training script first.')
````

## 6. Test Set Results

````python
results_path = '../results/textcnn_results.json'

if os.path.exists(results_path):
    with open(results_path) as f:
        results = json.load(f)

    print('TextCNN — Test Set Results')
    print('=' * 40)
    print(f"Accuracy       : {results['accuracy']}")
    print(f"Precision (M)  : {results['precision']}")
    print(f"Recall (M)     : {results['recall']}")
    print(f"F1-Score (M)   : {results['f1_score']}")
    print(f"Train Time (s) : {results['train_time_sec']}")
    print(f"Pred Time (s)  : {results['pred_time_sec']}")
else:
    print('Results not found. Run the training script first.')
````

## 7. Confusion Matrix

````python
cm_path = '../results/textcnn_confusion_matrix.png'

if os.path.exists(cm_path):
    img = plt.imread(cm_path)
    plt.figure(figsize=(7, 5))
    plt.imshow(img)
    plt.axis('off')
    plt.title('TextCNN — Confusion Matrix')
    plt.tight_layout()
    plt.show()
else:
    print('Confusion matrix not found. Run the training script first.')
````
