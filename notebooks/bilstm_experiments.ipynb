## 8. Attention Weight Visualisation

One advantage of using an attention mechanism is interpretability.
We can visualise which words the model focused on when making a prediction.

````python
def get_attention_weights(model, token_ids: torch.Tensor) -> np.ndarray:
    """Extract normalised attention weights for a single input."""
    model.eval()
    with torch.no_grad():
        embedded  = model.embedding(token_ids.unsqueeze(0).to(DEVICE))
        lstm_out, _ = model.bilstm(embedded)
        scores    = model.attention(lstm_out)          # (1, seq_len, 1)
        weights   = F.softmax(scores, dim=1).squeeze() # (seq_len,)
    return weights.cpu().numpy()


def visualise_attention(review_text: str, word2idx: dict,
                        model, max_seq_len: int = 200):
    """Plot attention weights over words in a review."""
    from utils.preprocessing import clean_text

    tokens  = clean_text(review_text).split()[:max_seq_len]
    indices = [word2idx.get(t, 1) for t in tokens]
    padding = [0] * (max_seq_len - len(indices))
    ids     = torch.tensor(indices + padding, dtype=torch.long)

    weights = get_attention_weights(model, ids)[:len(tokens)]

    plt.figure(figsize=(max(8, len(tokens) * 0.5), 2))
    plt.bar(range(len(tokens)), weights, color='purple', alpha=0.7)
    plt.xticks(range(len(tokens)), tokens, rotation=45, ha='right', fontsize=9)
    plt.title('BiLSTM Attention Weights')
    plt.ylabel('Attention Score')
    plt.tight_layout()
    plt.show()


# Example usage — update review text and load trained model first
# model.load_state_dict(torch.load('../results/bilstm_best.pt', map_location=DEVICE))
# visualise_attention(
#     review_text = 'This product is absolutely amazing and great quality',
#     word2idx    = data['word2idx'],
#     model       = model,
# )
print('Attention visualisation function ready.')
print('Uncomment the last 4 lines above after training is complete.')
````
