## 8. Self-Attention Weight Visualisation

The Transformer learns which tokens attend to which other tokens.
We can visualise the attention weights from the first encoder layer
to understand what the model focuses on.

````python
def visualise_self_attention(review_text: str, word2idx: dict,
                              model, max_seq_len: int = 200, head: int = 0):
    """
    Visualise self-attention weights from the first Transformer encoder
    layer for a given review.

    Args:
        review_text : Raw review string.
        word2idx    : Vocabulary mapping from preprocessing.
        model       : Trained SentimentTransformer.
        max_seq_len : Max token length used during training.
        head        : Which attention head to visualise (0-indexed).
    """
    from utils.preprocessing import clean_text

    tokens  = clean_text(review_text).split()[:max_seq_len]
    indices = [word2idx.get(t, 1) for t in tokens]
    padding = [0] * (max_seq_len - len(indices))
    ids     = torch.tensor([indices + padding], dtype=torch.long).to(DEVICE)

    model.eval()
    attn_weights = None

    # Hook to capture attention weights from the first encoder layer
    def hook_fn(module, input, output):
        nonlocal attn_weights
        # output[1] contains attention weights when need_weights=True
        if isinstance(output, tuple) and len(output) > 1:
            attn_weights = output[1]

    hook = model.transformer_encoder.layers[0].self_attn.register_forward_hook(hook_fn)

    with torch.no_grad():
        # Temporarily set need_weights=True
        model.transformer_encoder.layers[0].self_attn.need_weights = True
        _ = model(ids)
        model.transformer_encoder.layers[0].self_attn.need_weights = False

    hook.remove()

    if attn_weights is None:
        print('Could not extract attention weights. Try running after training.')
        return

    # Get weights for specified head, first sample, first (len+1) tokens (incl. CLS)
    n = len(tokens) + 1  # +1 for CLS
    labels = ['[CLS]'] + tokens

    w = attn_weights[0, :n, :n].cpu().numpy()  # (n, n)

    plt.figure(figsize=(max(6, n * 0.55), max(5, n * 0.5)))
    sns.heatmap(w, xticklabels=labels, yticklabels=labels,
                cmap='Oranges', annot=False)
    plt.title(f'Self-Attention Weights — Layer 1, Head {head}')
    plt.xticks(rotation=45, ha='right', fontsize=8)
    plt.yticks(rotation=0, fontsize=8)
    plt.tight_layout()
    plt.show()


# Example usage — uncomment after training is complete
# model.load_state_dict(torch.load('../results/transformer_best.pt', map_location=DEVICE))
# visualise_self_attention(
#     review_text = 'Battery life is terrible but the camera quality is amazing',
#     word2idx    = data['word2idx'],
#     model       = model,
# )
print('Self-attention visualisation function ready.')
print('Uncomment the last 4 lines above after training is complete.')
````
